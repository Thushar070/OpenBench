# Phase 12 — Neural Architecture Comparison

Review the saved validation-only comparison of Logistic Regression, TextCNN, vanilla RNN, LSTM, and GRU. This notebook reads saved artifacts; it does not train models or load the official test split.

```python
import json
from pathlib import Path
import pandas as pd

candidates = sorted(
    Path('outputs/neural_comparison').glob('imdb-neural-comparison-*/comparison.json'),
    key=lambda path: path.parent.name,
)
if not candidates:
    raise FileNotFoundError('Run `python -m deeptext.experiments.neural_comparison` first.')
comparison_path = candidates[-1]
comparison = json.loads(comparison_path.read_text(encoding='utf-8'))
assert comparison['validation_only'] is True
assert comparison['official_test_evaluated'] is False
assert comparison['sample_count'] == 5_000
comparison_path
```

```python
metrics = pd.DataFrame(comparison['metrics']).T[
    ['accuracy', 'precision', 'recall', 'f1', 'false_positive', 'false_negative']
].sort_values('f1', ascending=False)
metrics.style.format({
    'accuracy': '{:.2%}', 'precision': '{:.2%}',
    'recall': '{:.2%}', 'f1': '{:.2%}',
    'false_positive': '{:.0f}', 'false_negative': '{:.0f}',
})
```

```python
profiles = pd.DataFrame(comparison['model_profiles']).T
profiles[['parameter_count', 'training_duration_seconds', 'training_device']]
```

The official IMDb test split is reserved for final reporting. Validation metrics are a single seeded split and should not be interpreted as proof of generalization or statistical significance. See `comparison['pairwise_error_categories']`, `comparison['artifact_sizes']`, and `comparison['benchmarks']` for detailed records.
